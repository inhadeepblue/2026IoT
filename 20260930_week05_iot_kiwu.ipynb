```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
# from sklearn.linear_model import LinearRegression  # 선형회귀 모델
from sklearn.neighbors import KNeighborsRegressor  # K 최근접 이웃 모델

# csv 파일 다운로드 학습 준비
ls = pd.read_csv("https://github.com/ageron/data/raw/main/lifesat/lifesat.csv")
X = ls[["GDP per capita (USD)"]].values
y = ls[["Life satisfaction"]].values

# print(X)
# print(y)
# print(ls.info())
# print(ls.head(3))
# print(ls.tail(3))
# print(ls)
```

```python
# 데이터 시각화
ls.plot(kind='scatter', grid=True, x="GDP per capita (USD)", y="Life satisfaction")
plt.axis([23500, 62500, 4, 9])
plt.show()
```

```python
# 선형 회귀 모델 선택
# model = LinearRegression()

# KNN 모델 선택
model = KNeighborsRegressor(n_neighbors=3)

# 모델 학습
model.fit(X, y)

# 키프로스의 삶의 만족도 계산
X_new = [[37_655.2]] # 2020년 키프로스 GDP
print(model.predict(X_new)) # output: [[6.30165767]]
```
