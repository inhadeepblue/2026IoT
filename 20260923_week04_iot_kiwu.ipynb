```python
import seaborn as sns

df2 = sns.load_dataset('mpg')
print(df2[df2['horsepower'].isna()])
```

```python
print(df2.iloc[:,[1,3]])
```

```python
# print(df2.groupby('cylinders')['horsepower'].mean())
print(df2.groupby('cylinders')['horsepower'].agg(['mean','median','max','count']))
```

```python
df3 = sns.load_dataset('titanic')

print(df3.info())
print(df3.tail(10))
```

```python
print(df3.loc[:,['survived', 'alive', 'pclass', 'class', 'embarked', 'embark_town'] ])
```

```python
df4 = df3.drop(columns=['alive','class','embarked'])
print(df4.info())
```

```python
# print(df4['who'].value_counts())  # man / woman / child
# print(df4['sibsp'].value_counts())  # 0 ~ 8
# print(df4['parch'].value_counts())  # 0 ~ 6
# print(df4['adult_male'].value_counts())  # True, False
print(df4['deck'].value_counts())
```

```python
df4 = df4.drop(columns='deck')
print(df4.info())
```

```python
# print(df4.groupby('sex')['survived'].mean())
# print(df4.groupby(['sex','pclass'])['survived'].mean())
print(df4.groupby(['sex','pclass'])['survived'].agg(['mean', 'median', 'count']))
```
