```python
import pandas as pd

df = pd.read_csv('Uber.csv')
# print(df.info())
# print(df.head())
# print(df.tail())
print(df.describe())
```

```python
df['START_DATE*'] = pd.to_datetime(df['START_DATE*'], errors='coerce')
df['END_DATE*'] = pd.to_datetime(df['END_DATE*'], errors='coerce')
df = df.sort_values(['START_DATE*','END_DATE*'])
```

```python
# print(df.tail())
df = df.drop(1155)  # 결측치와 이상치 값을 가진 행 제거
print(df.tail())
print(df.describe())
print(df['MILES*'].sum())  # 누적거리였음을 확인
```

```python
print(df.info())
```

```python
# 특성 추출 (기존 특성들의 조합을 이용해 더 유용한 특성을 생성)
df['DURATION*'] = (df['END_DATE*'] - df['START_DATE*']).dt.total_seconds() / 60
print(df.info())
print(df.tail())
```

기계학습의 주요 도전과제 (bad model, bad data)
bad data
1. 불충분한 양의 훈련 데이터
2. 대표성이 없는 훈련 데이터
(삶의 만족도 예제의 27개국가에는 $23,500 이하 또는 $62,500 이상의 국가가 빠져있어
각 나라 삶의 만족도 모델 적용시 대표성이 떨어짐)
3. 표본추출 편향
4. 질낮은 데이터 (에러 범벅 데이터, 이상치가 가득한 데이터, 노이즈 많이 낀 데이터 등)
5. 무관한 특성 (예측하려는 타겟 값에 대해 아무런 유용한 정보나 예측 신호를 제공하지 않는 입력 변수)
특성공학
- 특성 선택 (가장 유용한 특성을 선택한다)
- 특성 추출 (기존 특성들의 조합을 이용해 더 유용한 특성을 생성)
- 특성 생성 (새로운 데이터를 수집해서 새 특성 생성)
