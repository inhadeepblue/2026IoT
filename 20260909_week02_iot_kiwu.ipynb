```python
import numpy as np

list_python = [9, '3', -7.9]
# for n in list_python:
#   print(n, type(n))


# 넘파이 배열 생성
array_numpy = np.array([9, '3', -7.9])
print(list_python)
print(array_numpy)
```

```python
a1 = np.zeros((3, 4))
# a2 = np.ones((2, 3, 5), dtype=int)
a2 = np.ones((2, 3, 5))
a3 = np.arange(5, 15, 3)
a4 = np.linspace(0, 1, 5)

# 넘파이 배열 정보
print(array_numpy.shape, array_numpy.ndim, array_numpy.dtype, array_numpy.size)
print(a1, a1.shape, a1.ndim, a1.dtype, a1.size)
print(a2, a2.shape, a2.ndim, a2.dtype, a2.size)
print(a3, a3.dtype)
print(a4)
```

```python
# 넘파이 배열 인덱싱 & 슬라이싱

a2[1, 0, 3] = 3.7
a2[0, 2, 1] = 9.1
print(a2)
```

```python
print(a4)
print(a4[1:4])
a1[0, 0] = 3
a1[2, 1] = 4
a1[1, 3] = 1
print(a1)
print(a1[1:,1:])
print(a1[:,1:3])
print(a1[:,[0, 3]])
```

```python
print(a1)

# 2차원 배열 각 원소에 3을 곱한다. 넘파이 스타일
a1 = a1 * 3
print(a1)

# 2차원 배열 각 원소에 3을 곱한다. 일반 파이썬 반복문 스타일
for i in range(3):
  for j in range(4):
    a1[i, j] = a1[i, j] * 3
print(a1)
```

```python
a5 = np.array([[3, 1, 2], [11, -9, 7]])
a6 = np.array([[2, 3, 6], [1, 9, 71]])
print(a5)
print(a6)
# print(a5 + a6)
a7 = a5 + a6
print(a7)
# 2차원 배열 원소중 9보다 작은 값을 1차원 배열로 만든다. 넘파이 스타일
print(a7[a7 < 9])

# 2차원 배열 원소중 9보다 작은 값을 1차원 배열로 만든다. 일반 파이썬 반복문 스타일
# temp = []
# for i in range(2):
#   for j in range(3):
#     if a7[i, j] < 9:
#       temp.append(int(a7[i, j]))
# print(temp)
```

```python
print(np.sum(a6))
print(np.min(a6))
print(np.max(a6))
print(np.mean(a6))
print(a6)
print(a6.reshape(3, 2))
print(a6.T)  # 전치행렬
```

```python
a8 = np.arange(10, 1, -1)  # 1차원 배열 생성 [10, 9, .... ,3 , 2]
print(a8)
a8 = a8.reshape(3,3)  # 1차원 배열을 2차원 배열로 형태 변환
print(a8)
a8 = a8.T  # 3x3 2차원 배열을 전치행렬로 변환
print(a8)
a8 = a8.flatten()  # 1차원으로 평탄화(펼치기)
print(a8)
```

```python
import pandas as pd

subject_dict = {'국':[99, 100, 89, 93], '영':[97, 91, 88, 92], '수':[100, 86, 94, 99]}
# print(subject_dict)
df1 = pd.DataFrame(subject_dict, index=[1, 2, 3, 4])
print(df1)
```
